# Generate a local synthetic dataset
This first example uses a sampler and does not send data to an external provider.

```python
import data_designer.config as dd
from data_designer.interface import DataDesigner

builder = dd.DataDesignerConfigBuilder()
builder.add_column(dd.SamplerColumnConfig(
    name="category",
    sampler_type=dd.SamplerType.CATEGORY,
    params=dd.CategorySamplerParams(values=["hardware", "software", "research"]),
))
result = DataDesigner().preview(config_builder=builder)
result.display_sample_record()
```

## Generate text with the active Cloudless model
The workbench can use whichever model is currently loaded in Model Manager. The endpoint stays on CloudlessOS's private container network.

```python
import os
import data_designer.config as dd
from data_designer.interface import DataDesigner

cloudless_provider = dd.ModelProvider(
    name="cloudless",
    endpoint=os.environ["OPENAI_BASE_URL"],
    provider_type="openai",
    api_key="OPENAI_API_KEY",
)
cloudless_model = dd.ModelConfig(
    alias="cloudless", model="cloudless", provider="cloudless",
    inference_parameters=dd.ChatCompletionInferenceParams(max_tokens=512),
)
builder = dd.DataDesignerConfigBuilder(model_configs=[cloudless_model])
builder.add_column(dd.SamplerColumnConfig(
    name="topic", sampler_type=dd.SamplerType.CATEGORY,
    params=dd.CategorySamplerParams(values=["robotics", "local AI", "synthetic data"]),
))
builder.add_column(dd.LLMTextColumnConfig(
    name="summary", model_alias="cloudless",
    prompt="Explain {{ topic }} in one concise sentence.",
))
result = DataDesigner(model_providers=[cloudless_provider]).preview(config_builder=builder)
result.display_sample_record()
```
